```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('processed_movies (2).csv')
```

```python
df.head()
```

```python
df['details'] = df['details'].fillna(' ')
df['genres'] = df['genres'].fillna(' ')
```

# **Text Vectorization**

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer(max_features=50000,ngram_range=(1,2))
```

```python
tfidf_matrix = tfidf.fit_transform(df['details'])
```

# **Calculating Cosine Similarity**

After converting movie descriptions into TF-IDF vectors, cosine similarity
is used to measure the similarity between movies.

The cosine similarity score ranges from -1 to 1. Since TF-IDF vectors contain
non-negative values, movie similarity scores will generally range from 0 to 1.

A higher score indicates that two movies have more similar textual features.

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def get_recommendations(title,n=10):
  if title not in df['title'].values:
    return ['Movie not found']

  idx = df.index[df['title'].values==title][0]

  sim_scores = cosine_similarity(tfidf_matrix[idx],tfidf_matrix).flatten()
  top_indices = sim_scores.argsort()[::-1][1:n+1]
  return df['title'].iloc[top_indices]
```

```python
get_recommendations('Toy Story', 5)
```

```python
get_recommendations('Jumanji', 5)
```

```python
get_recommendations('Avatar', 5)
```

```python
import pickle

pickle.dump(tfidf_matrix,open('tfidf_matrix.pkl','wb'))
df.to_pickle('df.pkl')
pickle.dump(tfidf,open('tfidf.pkl','wb'))
```
